## 1. Shoreline change
The Bruun rule shoreline change rate $\Delta y_{Bruun}$ is computed as:
$$
\Delta y_{Bruun} = \frac{c}{\tan \beta} \Delta  S
$$

where $\tan \beta$ is the active profile (depth of closure to dune crest) slope, and the constant $c$ is set to 1. 


### Data sources

- [CoastSat repository](https://uoa-eresearch.github.io/CoastSat/) for beach shoreface slopes ($\tan \beta _{shoreface}$) **
- [SLR scenarios](https://zenodo.org/records/11398538) from the NZ Sea Rise programme ($\Delta S$ ).

###

** To account for the shoreface being steeper than the active profile slope, an adjustment is applied when using the Bruun rule. Specifically, the shoreface slope is multiplied by a factor of 0.5, which is in line with LIDAR observations.



## 2. Reference year

The reference year is 2005, for both the sea level rise and the shoreline change projections.

## 3. Processing data

Processed and smoothed satellite-derived historic shoreline trends using the García et al. filter.

Processed  slope data by applying a Butterworth filter and adding small perturbations (white noise) to improve spatial variability.

## 4. Uncertainty bands

Uncertainty bands are quartiles of the SLR when used as inputs to calculate the Bruunian change.
